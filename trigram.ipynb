```python
import pandas as pd
import torch
```

```python
df = pd.read_csv('worldcities.csv')
city_names = df['city_ascii'].dropna().tolist()
print(f"Loaded {len(city_names)} city names")
city_names[:10]
```

```python
special_token = '<s>'
vocab = [special_token] + sorted(set(''.join(city_names)))
stoi = {s: i for i, s in enumerate(vocab)}
itos = {i: s for s, i in stoi.items()}
```

```python
N = torch.zeros(len(vocab), len(vocab), len(vocab), dtype=torch.int32)
for name in city_names:
    chars = [special_token, special_token] + list(name) + [special_token]
    for ch1, ch2, ch3 in zip(chars, chars[1:], chars[2:]):
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        ix3 = stoi[ch3]
        N[ix1, ix2, ix3] += 1
```

```python
P = N.float()
P /= P.sum(2, keepdim=True)
P = torch.nan_to_num(P)
```

```python
for i in range(10):
    out = ''
    ix1 = 0
    ix2 = 0
    while True:
        ix = torch.multinomial(P[ix1, ix2], num_samples=1, replacement=True).item()
        if ix == 0:
            break
        out += itos[ix]
        ix1 = ix2
        ix2 = ix
    print(out)
```
