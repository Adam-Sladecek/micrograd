```python
import pandas as pd
import torch
```

```python
df = pd.read_csv('worldcities.csv')
city_names = df['city_ascii'].dropna().tolist()
print(f"Loaded {len(city_names)} city names")
city_names[:10]
```

```python
special_token = '<s>'
vocab = [special_token] + sorted(set(''.join(city_names)))
stoi = {s: i for i, s in enumerate(vocab)}
itos = {i: s for s, i in stoi.items()}
stoi
```

```python
N = torch.zeros(len(vocab), len(vocab), dtype=torch.int32)
for name in city_names:
    chars = [special_token] + list(name) + [special_token]
    for ch1, ch2 in zip(chars, chars[1:]):
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        N[ix1, ix2] += 1
```

```python
P = N.float()
P /= P.sum(1, keepdim=True)
N
```

```python
for i in range(10):
    out = ''
    ix = 0
    while True:
        ix = torch.multinomial(P[ix], num_samples=1, replacement=True).item()
        if ix == 0:
            break
        out += itos[ix]
    print(out)
```
